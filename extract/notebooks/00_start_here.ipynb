Run these cells from top to bottom. The example clips are generated locally with FFmpeg. They contain a moving pattern with a tone and a short gray clip without audio. For transcription, use a video containing speech from your own collection. Change the settings below for your own collection.

```python
from pathlib import Path
import json
import pandas as pd
from mavic_extract import Project

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
videos = root / "example-videos"
# Generate synthetic clips for the first run. Comment out these two lines for your own collection.
from mavic_extract.examples import create_examples
create_examples(videos)
results = root / "results"
```

```python
project = Project.inspect(videos, results)
(root / ".mavic-project.json").write_text(json.dumps({"results": str(project.folder)}))
pd.DataFrame(project.manifest)[["video_id","duration_seconds","error"]]
```
