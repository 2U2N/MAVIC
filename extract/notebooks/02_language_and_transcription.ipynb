# Screen language and transcribe speech
Whisper large (screening) and turbo (transcription) each require a substantial first download and can run slowly on CPU. Select your device explicitly. Screening uses up to the first 30 seconds. Inspect the scores before starting full transcription.

```python
from pathlib import Path
import json
import pandas as pd
from mavic_extract import Project

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
project = Project(json.loads((root / ".mavic-project.json").read_text())["results"])
```

```python
device = "cpu"  # Use "cuda" for a configured CUDA GPU.
```

```python
project.run("language", device=device)
```

```python
pd.DataFrame([{"video_id": v["video_id"], **project.record(v,"language").get("values",{}), **project.record(v,"language").get("detail",{})} for v in project.manifest])
```

```python
project.run("transcription", device=device)
```

Only scores strictly above .98 qualify. The language score is a heuristic and can be confident even on audio without speech. Check the transcript results in the project records.
