# Measure transcripts and export tables
For a speech project run all cells. For audiovisual results only, run the project-loading cell and the export cells. Comment out any language models you do not need in the setup list.

```python
from pathlib import Path
import json
import pandas as pd
from mavic_extract import Project

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
project = Project(json.loads((root / ".mavic-project.json").read_text())["results"])
```

```python
from mavic_extract.linguistic import install_models
models = [
    "ca_core_news_md",
    "zh_core_web_md",
    "hr_core_news_md",
    "da_core_news_md",
    "nl_core_news_md",
    "en_core_web_md",
    "fi_core_news_md",
    "fr_core_news_md",
    "de_core_news_md",
    "el_core_news_md",
    "it_core_news_md",
    "ja_core_news_md",
    "ko_core_news_md",
    "lt_core_news_md",
    "mk_core_news_md",
    "nb_core_news_md",
    "pl_core_news_md",
    "pt_core_news_md",
    "ro_core_news_md",
    "ru_core_news_md",
    "sl_core_news_md",
    "es_core_news_md",
    "sv_core_news_md",
    "uk_core_news_md",
]
install_models(models)
```

```python
project.run("text", models=models)
```

```python
from mavic_extract.export import export
purpose = "both"  # "fixed", "candidate", or "both"
export(project, purpose)
```
